# Heart Disease Predictions

## Import Libraries

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

## Fetch & Explore the data

```python
df = pd.read_csv('framingham.csv')
df.shape
```

```python
df.head()
```

```python
df.isnull().sum(axis=0)
```

```python
df.dropna(inplace=True)
```

```python
df.TenYearCHD.value_counts()
```

```python
# We have impalance data that ~= 85% didn't has risk for CHD and ~= 15 has risk
plt.figure(figsize=(16,8))
plt.subplot(1,2,1)
plt.pie(df.TenYearCHD.value_counts(),autopct='%1.1f%%',labels=['Not Risk','Risk'])
plt.subplot(1,2,2)
ax = sns.countplot(df,x='TenYearCHD',hue='TenYearCHD')
ax.bar_label(ax.containers[0],padding=3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16,20))
index = 1
for col in df.columns:
    plt.subplot(8,2,index)
    sns.histplot(df[col],kde=True)
    plt.title(col)
    index +=1
plt.tight_layout()
plt.show()
```

```python
X = df.drop(columns=['TenYearCHD'])
y = df.TenYearCHD
transform_cols = ['age', 'cigsPerDay', 'totChol', 'sysBP', 'diaBP', 'BMI', 'heartRate', 'glucose']
X_scaled = X.copy()
X_scaled[transform_cols] = StandardScaler().fit_transform(X_scaled[transform_cols])
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_scaled,y,test_size=0.3,random_state=44)
print(X_train.shape,X_test.shape,y_train.shape,y_test.shape)
```

## Make and Evaluate the Model

```python
from sklearn.linear_model import LogisticRegression

logModel = LogisticRegression( max_iter=1000)
logModel.fit(X_train,y_train)

preds = logModel.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print('The details for confusion matrix is =')
print (classification_report(y_test, preds))

cm = confusion_matrix(y_test, preds)
conf_matrix = pd.DataFrame(data = cm, 
                           columns = ['Predicted:0', 'Predicted:1'], 
                           index =['Actual:0', 'Actual:1'])

plt.figure(figsize = (8, 5))
sns.heatmap(conf_matrix, annot = True, fmt = 'd', cmap = "coolwarm")

plt.show()
```

* We Noticed that the model good to predict the healty people (0 - class) but perform poorly in the patients (1 - class)
* In this dataset to predicted the patient as a healty one is more dangerous that predict the healty one to patient, so we have a problem with the FN (Recall)
* In future we should solve the high imbalanced of the data that the model can get the pattern of the patients
* We should try the **Downsampling** & **Upweight**
